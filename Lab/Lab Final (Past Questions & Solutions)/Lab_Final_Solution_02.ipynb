# **Question-01**

```python
g=Polynomial([-9, 3, -7, 1, 0, -2])
#a
print(3*g(-1)+10)
#b
x=1
print(2*g.deriv(3)(x)-g.deriv(1)(x)/3)
l=np.linspace(-4, 5, 500)
plt.plot(l, g(l), color='red')
plt.plot(l, g.deriv(2)(l), color='green')
plt.show()
```

# **Question-02**

```python
def l_k(data_x, k, x):
  lk=1
  for i in range (len(data_x)):
    if i%2!=0 and i!=k:
      lk*=(x-data_x[i])/(data_x[k]-data_x[i])
  return lk

def get_poly_lagrange(data_x, data_y, x_arr):
  res=[]
  for i in range (len(x_arr)):
    temp=0
    for j in range (len(data_y)):
      temp+=data_y[-(j+1)]*l_k(data_x, j, x_arr[i])
    res.append(temp)
  return res

data_x=np.array([1, 3, 6, 10])
data_y=np.array([4, 9, 12, 20])
x_arr=np.array([0, 2, 5, 8])
print(get_poly_lagrange(data_x, data_y, x_arr))
```
