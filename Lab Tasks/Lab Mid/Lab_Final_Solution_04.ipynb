# **Question-01**

```python
#a
g=Polynomial([7, -5, 4, 0, -2, 3])
print(g)
print(g.degree())
#b
print(3*g(1.5)-7)
#c
x=-1
print(2*g.deriv(3)(x)-(g.deriv(1)(x)/3))
#d
l=np.linspace(-5,4,10)
plt.plot(l, g(l), color='green', label="g(x)")
plt.plot(l, g.deriv(2)(l), color='red',label="g''(x)")
plt.legend()
plt.show()
```

# **Question-02**

```python
#a
def l_k(data_x, k, x):
  lk=1
  for j in range(len(data_x)):
    if j!=k and j%2==0:
      lk*=(x-data_x[j])/(data_x[k]-data_x[j])
  return lk

#b
def get_poly_lagrange(data_x, data_y, x_arr):
  res=[]
  for i in range (len(x_arr)):
    temp=0
    for j in range (len(data_x)):
      temp+=data_y[-(j+1)]*l_k(data_x,j, x_arr[i])
    res.append(temp.round(8))
  return res
data_x=np.array([1, 3, 6, 10])
data_y=np.array([4, 9, 12, 20])
x_arr=np.array([0, 2,5,8])

print(get_poly_lagrange(data_x, data_y, x_arr))
```

# **Question-03**

```python
f=Polynomial([5, -3, 2, 1])
x=2
h=0.8
t=0.1
c=0

def fwd (f, x, h):
  return (f(x+h)-f(x))/h

while True:
  error=abs(f.deriv(1)(x)-fwd(f,x,h))

  if error<t:
    break
  h=h/4
  c+=1

print("Required h:", h)
print("Total number of reductions:",c)
```
