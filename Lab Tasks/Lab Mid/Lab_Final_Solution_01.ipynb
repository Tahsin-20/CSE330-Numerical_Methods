## **Question-03**

```python
x_data=[1, 3, 5]
y_data=[2, 12, 30]

def l(k, x, x_data):
  l_k=1
  for i in range (len(x_data)):
    if i!=k:
      l_k*=(x_data[k]-x_data[i])/(x_data[k]-x)
  return l_k

lagrange_bases=[]
for i in range (len(x_data)):
  lagrange_bases.append(l(i,7,x_data))
print(lagrange_bases)

p=0
for i in range (len(lagrange_bases)):
  p+=lagrange_bases[i]*y_data[i]
print(p)
```

## **Question-04**

```python
g=Polynomial([8, 4, -5, 2])
x=2
h=0.5
g_prime=g.deriv(1)


def fwd_diff(x, h):
  return (g(x+h)-g(x))/h

while True:
  error=abs(g_prime(x)-fwd_diff(x,h))

  if error<0.0015:
    break
  h=h/4

print(h)
print(error)
```
