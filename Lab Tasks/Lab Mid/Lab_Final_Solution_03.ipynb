# **Question-02**

```python
def l_k(data_x, k, x):
  lk=1
  for i in range (len(data_x)):
    if i!=k:
      lk*=(data_x[k]-data_x[i])/(x-data_x[i])
  return lk

def hk (data_x, k, x):
  return (2*x+4*(x**3-data_x[k])*l_k(data_x,k,x)**2)

def h_hat (data_x, k, x):
  lk=l_k(data_x, k, x)
  return ((x**2-data_x[k])*lk-x+3*x**3)

data_x=np.array([5, 13, 22, 80])
print(l_k(data_x, 2, 2))
print(hk(data_x, 2, 2))
print(h_hat(data_x, 2, 2))
```

# **Question-03**

```python
f=Polynomial([6, 3, 7, 0, -3, 3])
def fwd(f, x, h):
  return (f(x+h)-f(x))/h
h=1.5
c=0
while True:
  error=abs(f.deriv(1)(3)-fwd(f, 3, h))

  if error<0.002463:
    break
  h=h/3
  c+=1
print(h)
print(c)
```
